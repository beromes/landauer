```python
import landauer.parse as parse
import landauer.entropy as entropy
import landauer.evaluate as evaluate
import landauer.graph as graph

half_adder = '''
    module example (a, b, x, y);
        input a, b;
        output x, y;

        assign x = a & b;
        assign y = a & b;
    endmodule
'''
aig = parse.parse(half_adder)
entropy_database = entropy.entropy(aig)
evaluation = evaluate.evaluate(aig, entropy_database)
graph.show(graph.default(aig))
```

```python
import landauer.summary as summary
import landauer.algorithms.genetic as genetic

param_map = {
    'name': 'Parametrização Básica',
    'n_generations': 5,
    'n_initial_individuals': 5,
    'reproduction_rate': 1,
    'mutation_rate': 0.2,
    'mutation_intensity': 0.1,
    'elitism_rate': 0.1
}


results = genetic.genetic(aig, entropy_database, param_map, show_debug_messages=False, timeout=300, plot_circuit=True)
```
